```python
import numpy as np
```

```python
def print_pi_estimate(pi_estimate):
  relative_error = np.abs(np.pi - pi_estimate) / np.pi
  print(f"pi is approximately: {pi_estimate:.5f} (relative_error: {relative_error:.3%})")
```

Method 1: Monte-Carlo Integration. We all know (hopefully) that the area of a square is $s^2$ and the area of a circle is $\pi r^2$. If we inscribe a unit circle (a circle with radius 1) inside a square with side lengths of 2, the ratio of their areas will be $\frac{\pi}{4}$. If we treat the square like a dartboard and throw a dart at a random point within the square, the probability that it lands inside the circle will also be $\frac{\pi}{4}$. We can estimate this probability by throwing many, many darts at the square and counting the fraction that fall inside the circle.
$$
\begin{align*}
  \frac{\text{hits}}{\text{total}} &\approx \frac{\pi}{4}\\
  \pi &\approx 4 \times \frac{\text{hits}}{\text{total}}
\end{align*}
$$

```python
def method_1(n_trials):
  '''
  Estimate pi using Monte-Carlo Integration (method 1). Throw n_trials "darts" and estimate pi using the method described above.
  '''
  coords = np.random.uniform(-1, 1, (n_trials, 2))
  inside_circle = np.sqrt(coords[:, 0]**2 + coords[:, 1]**2) < 1
  return 4 * np.mean(inside_circle)

print_pi_estimate(method_1(10000))
```

Method 2: Coprimality. The probability that two randomly chosen natural numbers are coprime is $\frac{6}{\pi^2}$. Curious where this result comes from?

The probability that a random natural number is divisible by a prime $p$ is $\frac{1}{p}$ because multiples of $p$ are spaced $p$ numbers apart. So the probability that two independently chosen natural numbers are *both* divisible by $p$ is $\frac{1}{p^2}$. The probability that one or both are *not* divisible by $p$ is then given by the complement, $1 - \frac{1}{p^2}$. For two numbers to be coprime, this must hold for all primes $p$. The probability of this happening is given by the infinite product
$$
\begin{align*}
&= \Pi_{p \in \mathbb{P}} \left(1 - \frac{1}{p^2}\right)\\
&= \sum_{n=1}^\infty \frac{1}{n^2} & \text{By Euler's Product Formula}\\
&= \zeta(2)\\
&= \frac{6}{\pi^2} & \text{Known Value of the Riemann-Zeta Function}
\end{align*}
$$

Similar to the method above, generate a bunch of random integers *pairs*, and the fraction of them that are coprime serves as an unbiased estimate of the true probability $\frac{6}{\pi^2}$. Using some basic algebra (as above) you can solve for an estimated value of $\pi$.

```python
def method_2(n_trials):
  # TODO
  # NOTE: high can be any large number, technically the math is designed for arbitrarily large numbers but we have to put a limit somewhere
  naturals = np.random.randint(1, 1000000, (n_trials, 2))
  are_coprime = np.gcd(naturals[:, 0], naturals[:, 1]) == 1
  return np.sqrt(6 / np.mean(are_coprime))

print_pi_estimate(method_2(1000000))
```

Method 3: Madhava-Leibiniz Series.
$$
\begin{align*}
  \arctan(1) &= \frac{\pi}{4}\\
  \pi &= 4\arctan(1)\\
  &= 4\sum_{n=0}^\infty (-1)^n \frac{1^{2n + 1}}{2n + 1}\\
  &= 4\sum_{n=0}^\infty \frac{(-1)^n}{2n + 1}\\
  &\approx 4\sum_{n=0}^N \frac{(-1)^n}{2n + 1}
\end{align*}
$$

We can use this series to get the value of $\pi$ to arbitrarily good precision by adding more terms unlike Methods 1 and 2 (Monte Carlo methods). Try doing this **WITHOUT ANY LOOPS!**

```python
def method_3(n_terms):
  indices = np.arange(n_terms)
  return 4 * np.sum((-1)**indices / (2 * indices + 1))

print_pi_estimate(method_3(1000))
```
